- longitude = decimal degrees [-360,360]
- depth = depth of the event (km)
- mag =  magnitude of event

# 7

a. Create a new variable `top20` with the 20 earthquakes with largest magnitude. HINT: check the method [`pandas.Series.nlargest`](https://pandas.pydata.org/docs/reference/api/pandas.Series.nlargest.html). Your answer should contain only the `id` and `mag` columns. 

```python
# Filter the eqk dataset to include only earthquakes where the type is equal to earthquake.
# Then use nlargest to ask for the 20 largest values in the mag column, and include only the mag column
# (ID is included as the index).
top20 = eqk[eqk['type']=='earthquake'].nlargest(20, 'mag')['mag']
```

```python
# View the dataframe 
top20
```

# 8 
We are interested in the place, magnitude, and depth of the top 20 earthquakes with largest magnitude. Select this data from the `eqk` data frame. Store your answer in the `top20_filtered` variable. 

HINT: This is a "select rows and columns simultaneously" from `eqk` exercise. To select the rows you could use the index of `top20`. Make sure your columns are in the following order to pass the test: `place`, `mag`, `depth`. 

```python
# Create a dataframe that includes the place, mag, and depth columns from eqk. Use nlargest and .index to find
# the location of the largest 20 mags, and loc to pull the whole row.
top20_filtered = eqk[['place', 'mag', 'depth']].loc[eqk['mag'].nlargest(n=20).index]
```

# 9 

Without creating any new variables, visualize the locations of eartquakes with magnitude greter than 5 by creating a scatter plot of their latitude and longitude. Make the following adjustments to the basic plot:
- update the title
- color the points by magnitude (HINT: [check the `c` parameter here](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.plot.scatter.html))
- adjust the `alpha` (transparency) argument
- adjust the `colormap` argument to maket it look prettier. [Click for ideas of sequential colormaps.](https://matplotlib.org/stable/users/explain/colors/colormaps.html)


```python
# Filter the eqk dataframe to where the mag column has values greater than 5. 
# Create a scatter plot out of these values, using x and y for long and lat respectively, and coloring the
# points by mag. 
eqk[eqk['mag']>5].plot(kind='scatter',
                      x='longitude', y='latitude',   # Set the variables to be used for X and 
                      c='mag',   # Set a variable to color the points
                      alpha= .4,   # Adjust point transparency
                      colormap='plasma')  # Change color scheme
```
